```python
bruto = pd.read_excel(path, sheet_name="Municípios", header=None)
bruto
```

```python
pop = bruto.iloc[3:5573].copy()
pop
```

```python
pop.columns = [
    "col_vazia",
    "uf",
    "cod_uf",
    "cod_munic",
    "municipio",
    "pop_2010_sinopse",
    "pop_2010_compat",
    "pop_2022",
]

pop = pop.drop(columns=["col_vazia"])
pop = pop.reset_index(drop=True)
pop
```

```python
for col in ["pop_2010_sinopse", "pop_2010_compat", "pop_2022"]:
    pop[col] = pd.to_numeric(pop[col], errors="coerce")

pop.describe().round(2)
```

```python
pop[pop["uf"].isna() | pop["pop_2022"].isna()]
```

```python
pop_estado = pop.groupby("uf", as_index=False)[["pop_2010_compat", "pop_2022"]].sum()
pop_estado
```

```python
pop_estado["diferenca"] = pop_estado["pop_2022"] - pop_estado["pop_2010_compat"]
pop_estado["cresc_pct"] = (pop_estado["diferenca"] / pop_estado["pop_2010_compat"] * 100).round(2)

pop_estado
```

```python
pop_estado = pop_estado.sort_values("diferenca", ascending=False).reset_index(drop=True)
pop_estado
```

```python
pop_estado.to_csv("populacao_por_estado_2010_2022.csv", index=False, encoding="utf-8-sig")
print("salvo: populacao_por_estado_2010_2022.csv")
```

```python
pop_municipio = pop[["uf", "cod_uf", "cod_munic", "municipio", "pop_2010_compat", "pop_2022"]].copy()
pop_municipio
```

```python
pop_municipio["diferenca"] = pop_municipio["pop_2022"] - pop_municipio["pop_2010_compat"]
pop_municipio["cresc_pct"] = (pop_municipio["diferenca"] / pop_municipio["pop_2010_compat"] * 100).round(2)

pop_municipio
```

```python
pop_municipio = pop_municipio.sort_values("diferenca", ascending=False).reset_index(drop=True)
pop_municipio.head(20)
```
